## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is a **ranking/scoring** problem.

The decision is not simply whether a page is declining, but **which content pages should an SEO editor review first**. The model will assign a priority score so that pages with the highest likelihood of needing a content refresh appear at the top of the recommendation list.

This supports decision-making rather than replacing human judgment.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**Decision**

Which content pages should be refreshed first?

**Who acts on the output?**

An SEO analyst or content editor.

**Action**

Review and refresh the highest-priority pages before lower-priority pages.

**Cost of a wrong recommendation**

A false positive wastes editor time reviewing pages that did not need attention.

A false negative is more costly because an important declining page may continue losing search visibility.

For the starter dataset, the observed historical outcome (`trend_direction`) is used only as an evaluation proxy. The model's output is a priority score rather than a final business decision.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary evaluation metric is **Precision@50**.

Precision@50 measures how many of the top 50 recommended pages are actually observed to be declining.

This metric matches the real decision because editors usually review only a limited number of pages rather than every page in the dataset.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule such as "refresh pages older than 180 days" considers only one signal.

Content performance depends on several interacting factors including impressions, click-through rate, average search position, content age, and update history. These relationships are difficult to describe with a single if-statement and may vary across different types of content.

A ranking model can combine multiple observed signals into a priority score that better supports editorial decisions. The model provides **decision support**, not automatic decisions, and the results should be interpreted as observed and directional rather than causal.

### Problem Frame

For SEO analysts deciding which content pages to refresh first, I will build a ranking model using historical content performance data to produce a priority score. The quality of the ranking will be measured using Precision@50. A wrong recommendation may waste editor time or delay updates to pages that are actually declining. A simple rule is insufficient because content performance depends on multiple interacting signals. This work provides decision support using observed historical data and does not claim to predict Google's ranking algorithm or establish causal relationships.
